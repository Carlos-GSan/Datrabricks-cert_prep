## Delta Live Tables (DLT)

Delta Live Tables is a **declarative ETL framework** for building reliable, maintainable, and testable data processing pipelines

You define the transformations to perform on your data and Delta Live Tables manages **task orchestration, cluster management, monitoring, data quality, and error handling**.

It handles both streaming and **batch workloads** with minimal manual intervention.

### Delta Live Tables(DTL) Architecture

Delta Live Tables is built on top of hte familiar Apache Spark and Delta Lake, a spark being the Compute engine, you can implement both streaming and batch workloads in Delta Live tables

Databricks identifies Delta notebooks or standard spark notebooks.

- You can only write the DLT code using SQL or Python.
- Use SQL for most use cases because the Delta syntax in SQL is more user friendly
- Python has more flexibility and its suitable for most complex worflows
- DLT notebooks usually contains the transformation logic as well as the definition of the data quality expectations.


The other part of user interface is the pipelines. 
- Pipelines brings all the notebooks together, and this is the place where you specify your infrastructure requirements

- some of the recent performance improvements to DLT are only available wht used Unity Catalog

![imagen_1791710838830.png](./imagen_1791710838830.png "imagen_1791710838830.png")

Starting a DLT pipeline doesn't immediately execute the notebooks. 
It first gathers all the information from the notebooks, and then it creates a DAG.

The cluster provisioned will be based on the definition from the pipeline.

It can either be serverless cluster or a job cluster

then the code will be executed during the execution

During the excecution it gathers various metadata such as pipeline status and health of the pipeline, metadata from the data quality expectations such as number of failures, success, etc, and also data lineage

### Delta Live Tables - Programming

#### Imperative

- **How to do a task?**: User provides step-by-step instruction

```
user_above_18 = []
for user in users:
    if user.age > 18:
        users_above_18.append(user)
```

#### Declarative

- **What should be done?**: System decides how to do it

```
SELECT * FROM users WHERE age > 18;
```

Delta Live uses declarative programming

- No need to handle checkpoints, fault tolerance, dependecies etc.
- No need to write complex logic such as CDC, Schema Evolution ect.
- Better Optimization
