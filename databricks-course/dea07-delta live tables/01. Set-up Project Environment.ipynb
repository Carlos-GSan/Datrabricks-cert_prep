## Set-up the project environment for CircuitBox Data Lakehouse

1. Create external location - dea_course_ext_dl_circuitbox
2. Create Catalog - circuitbox
3. Create Schemas
    - landing
    - lakehouse
4. Create Volume - operational_data

![imagen_1791916072734.png](./imagen_1791916072734.png "imagen_1791916072734.png")

#### 1. Create External Location

**External Location Name**: dea_course_ext_dl_circuitbox

ADLS Path: `abfss://circuitbox@deacourseext2.dfs.core.windows.net/`

*Storage Credential*: dea_course_ext_sc

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_circuitbos
    URL 'abfss://circuitbox@deacourseext2.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL dea_course_ext_sc)
    COMMENT 'External Location for the circuitbox data lakehouse';
```

```sql
%fs ls 'abfss://circuitbox@deacourseext2.dfs.core.windows.net/'
```

## 2. Create Catalog

**Catalog Name**: circuitbox

*Managed Location*: `abfss://circuitbox@deacourseext2.dfs.core.windows.net/`

```sql
CREATE CATALOG IF NOT EXISTS circuitbox
    MANAGED LOCATION 'abfss://circuitbox@deacourseext2.dfs.core.windows.net/'
    COMMENT 'Catalog for the circuitbox data lakehouse'
```

```sql
SHOW CATALOGS;
```

### 3. Create Schemas

1. **Schema Name**: Landing
*Managed Location*: `abfss://circuitbox@deacourseext2.dfs.core.windows.net/landing`

2. **Schema Name**: lakehouse
*Managed Location*: `abfss://circuitbox@deacourseext2.dfs.core.windows.net/lakehouse`

```sql
USE CATALOG circuitbox;

CREATE SCHEMA IF NOT EXISTS landing
    MANAGED LOCATION 'abfss://circuitbox@deacourseext2.dfs.core.windows.net/landing';

CREATE SCHEMA IF NOT EXISTS lakehouse
    MANAGED LOCATION 'abfss://circuitbox@deacourseext2.dfs.core.windows.net/lakehouse';
```

```sql
SHOW SCHEMAS;
```


### 4. Create Volume

**Volume Name**: operational_data

*ADLS Path*: `abfss://circuitbox@deacourseext2.dfs.core.windows.net/landing/operational_data/`

```sql
USE CATALOG circuitbox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://circuitbox@deacourseext2.dfs.core.windows.net/landing/operational_data';
```

```sql
%fs ls /Volumes/circuitbox/landing/operational_data
```
