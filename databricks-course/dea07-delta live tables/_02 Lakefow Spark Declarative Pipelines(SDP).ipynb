### Compute/Cluster Configuration

When you execute the DLT pipelines, a job cluster is automatically created and the DLT notebooks will be executed on that.

![imagen_1791832879663.png](./imagen_1791832879663.png "imagen_1791832879663.png")

#### Cluster configuration requirements
In order to work with Unity Catalog, DLT requires us to create the cluster with shared Access mode.
So it has to have at least one worker node

##### Job Compute
- Shared Access Mode
    - Multi Node Cluster
    - 8vCPUs

##### Interactive Compute
- Shared Access Mode
    - Multi Node Cluster
    - 8vCPUs

We can use the serverless compute instead of using another cluster(use that if thas available)
