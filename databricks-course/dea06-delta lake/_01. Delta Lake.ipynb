## What is Delta Lake?

Delta Lake is the **optimized storage layer** that provides the foundation for tables in Lakehouse on Databricks. Delta lake is open source software that exteds Parquet data files with a file-based transaction log for **ACID transactions** and **scalable metadata handling**.

It allows you to use a single copy of data for **both batch and streaming operations** and providing **incremental processing** at scale

#### ACID transactions
ACID stands for Atomicity, Consistency, Isolation and Durability. Concurrent transactions can be carried out without affecting the integrity of the data

- **Scalable Metadata**: Includes the data in the form of parquet files, as well metadata in format of JSON. Metadata in this context is basically the transaction log files that provides the ability to perform ACID transcations.
- **Time Travel**:With the help of transaction logs, Delta Lake offers the ability to query previous versions of the table. Especially useful when you're dealing with historical data issues, audits, data compliance issues, etc..
- **Simple Solution Architecture**: This is enabled by having a single spark API for both batch and streaming workloads
- **Support for DML Operations**: Delta Lake makes it easier to updating incorrect data by providing support for all DML operations such as insert, update, delete, merge
- **Better Performance**: The engine offers better performance on the queries executed in the data Lake house
- **Open Source**: It is easy to work with a wide variety of tools.

#### Architecture

Deltalake stores the data in Parquet files. Parquet is a columnar binary file format widely used in big data projects, and it's efficient for storing and querying large data sets while storing the data in parquet file format.

Delta Lake also creates a transaction log alongside that file.

### Delta Transaction Log

Databricks registers the table metadata in unity catalogo, and creates a folder in our cloud storage to keep the data files and transaction log files related to this table.
The transaction log file is created un the folder `_delta_log/`.

When we insert some data to the table, it may result in one or more parquet files depending on the amount of data returned.
As soon as the precoess completed, write in the parquet file. A new transaction log file will be created to indicate that we've added some data to the table. And the data is present in file one dot parquet file.

Then we insert two aditional files. So the precess writes two addtional data files containing the inserted data. But then a new transaction log file will be created to indicate that we've added two new data files as part of the transaction.
