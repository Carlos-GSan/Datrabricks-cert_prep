### 1. Difference between Create or Replace and Drop and Create Table statements

The main difference between create or replace and drop and create table statements is the way Delta Lake maintains history

#### Behaviour of the DROP and CREATE statements

```sql
DROP TABLE IF EXISTS demo.delta_lake.companies;

CREATE TABLE demo.delta_lake.companies
    ( company_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING,
    founded_date DATE,
    country STRING);

INSERT INTO demo.delta_lake.companies
(company_name, founded_date, country)
VALUES  ('Apple', '1976-04-01', 'USA'),
        ('Microsoft', '1975-04-01', 'USA'),
        ('Google', '1998-09-04', 'USA'),
        ('Amazon', '1994-05-05', 'USA');
```

```sql
DESC HISTORY demo.delta_lake.companies;
```

We los all the history as soon as we ran the drop table statement. And when created the table, we've got the history again

#### Behaviour of the CREATE OR REPLACE statement

```sql
DROP TABLE IF EXISTS demo.delta_lake.companies;
```

```sql

CREATE OR REPLACE TABLE demo.delta_lake.companies
    ( company_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING,
    founded_date DATE,
    country STRING);

INSERT INTO demo.delta_lake.companies
(company_name, founded_date, country)
VALUES  ('Apple', '1976-04-01', 'USA'),
        ('Microsoft', '1975-04-01', 'USA'),
        ('Google', '1998-09-04', 'USA'),
        ('Amazon', '1994-05-05', 'USA'),
        ('Tencent', '1998-11-21', 'China');
```

```sql
DESC HISTORY demo.delta_lake.companies;
```

The history is now been uptaded when create or replace the second time, doenst eliminate the history.

Use the `CREATE OR REPLACE` statement if you want to keep the history in delta lake tables

### 2. CTAS statement
CTAs are create table as select statment basically allows you to create a new table and populate it with the results of a Select query in a single operation.

#### Limitations
- You cannot specify column proerties.(not null contraints, column data types, etc)
- The column data types have been inferred.


In order to change the column type, we can cast the column type.
or we can use ALTER TABLE in order to change the properties

```sql
DROP TABLE IF EXISTS demo.delta_lake.companies_china
```

```sql
CREATE TABLE demo.delta_lake.companies_china
SELECT CAST(company_id AS INT) as company_id,
       company_name,
       founded_date,
       country
    FROM demo.delta_lake.companies
WHERE country = 'China'
```

```sql
select * from demo.delta_lake.companies_china
```

```sql
DESC HISTORY demo.delta_lake.companies_china
```

```sql
ALTER TABLE demo.delta_lake.companies_china
    ALTER COLUMN founded_date COMMENT 'Date the company was founded'
```

```sql
ALTER TABLE demo.delta_lake.companies_china
    ALTER COLUMN company_id SET NOT NULL;
```

```sql
DESC EXTENDED demo.delta_lake.companies_china;
```
