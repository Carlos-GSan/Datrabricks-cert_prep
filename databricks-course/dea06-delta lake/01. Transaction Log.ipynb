## Understanding Delta Lake Transaction Log
Understand the cloud storage directory structure behind delta lake tables

### 0. Create a new shema under the demo catalog for this section of the course(delta_lake)

```python
%sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
    MANAGED LOCATION 'abfss://demo@deacourseext2.dfs.core.windows.net/delta_lake'
```

### 1. Create a Delta Lake Table

```python
%sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.companies
    (company_name STRING,
    founded_date DATE,
    country STRING);
```

```python
%sql
DESC EXTENDED demo.delta_lake.companies
```

Every object in Unity Catalog is given a unique identifier

### 2. Insert some data

```python
%sql
INSERT INTO demo.delta_lake.companies
VALUES ("Apple", "1976-04-01", "USA");
```

```python
%sql
select * from demo.delta_lake.companies
```

This transaction generates one parque file in the cloud storage and a json in the _delta_log

```python
%sql
INSERT INTO demo.delta_lake.companies
VALUES  ("Microsoft", "1975-04-01", "USA"),
        ("Google", "1998-08-01", "USA"),
        ("Amazon", "1994-07-05", "USA");
```

```python
%sql
select * from demo.delta_lake.companies
```
