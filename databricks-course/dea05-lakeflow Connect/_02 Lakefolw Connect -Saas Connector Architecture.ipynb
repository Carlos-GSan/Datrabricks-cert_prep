## Saas Connectors
- Managed Connectors for both saas and databases

We can ingest data from Saas Applications like Salesforce, Workday and more ..., by pulling data via APIs.
So we need to create a connection in Databricks. This connetion is defined as a Unity Catalog object, its centraly managed and it can be used acroos different pipelines.

It's a good idea to create a dedicated user in Salesforce just for data ingestion. That way, you can control exactly what can be accessed. Or restrict access to specific objects or columns using permissions in Salesforce.

- Ingestion Pipeline: This is fully managed by Databricks and runs on a serverless compute.
- The data is returned to Delta tables in a bronze layer.

#### Change Data Capture (CDC)

Most databases has the ability to capture just the changes and write them to logs so that the ingestion gateway can read those changes from the logs and processes.

These are turned on and off by the database administrators of those databases
- CT(Change Tracking): Can only apply on a table with a primary key. (lightweight)
- CDC(Chage Data Capture): Use it if table doesn't have a primary key

If both CDC & CT are enabled: Change Tracking will be enabled

![imagen_1791399495658.png](./imagen_1791399495658.png "imagen_1791399495658.png")

Each component has a very specific role:

- Connection: responsible for keeping the credentials
- Ingestion gateway: Responsible for establishing the connection, identifying and capture the changes to the data conitinously
- Staging storage: responsible for holding the data safely and decoupling the process
- Ingestion pipeline is responsible for processing and loading the data into the data lake house.
