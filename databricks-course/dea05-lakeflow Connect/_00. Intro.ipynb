
## Autoloader

- Auto Loader is a new structured streaming source designed for large-scale, efficient data ingestion
- Auto Loader incrementally and effciently processes new data files as they arrive in the cloud storage

### Why we need another solution like autoloader?

The traditional File Stream Source has some limitations when processing large amount of data.
- Traditional file streaming sources repeatedly scan entire directories to detect new files
- Listing files from an entire folder is solw and expensive
- Duplicate file detection has **scalability issues**, large volumes of data can cause memory constraints.
- When working with traditional file streaming source, users must manually define the schema before the straming starts (**Schema Evolution Problems**)

**Auto Loader**
- Efficient File Detection using Cloud Services like aws S3 event notifications or Azure Event Grid to track new files.
- Reads from a cloud queue to look for new files for real time ingestion. (Scalability Improvements)
- Supports automatic schema inference from sample files and schema evolution by automatically adding new columns or rescuing unexpected data in a separate rescue data column
- offers different schema evolution modes allowing users to control how schema changes are handled

```
customers_df = spark.readStream
                    .format("cloudFiles")
                    .option("cloudFiles.format", "json")
                    .schema(customer_schema)
                    .load("/Volumes/gizmobox/landing/operational_data/customers_stream")
```

We can use file notification mode, which leverages cloud storages services to track new files.

```
customers_df = spark.readStream
                    .format("cloudFiles")
                    .option("cloudFiles.format", "json")
                    .option("cloudFiles.useNotifications", "true")
                    .schema(customer_schema)
                    .load("/Volumes/gizmobox/landing/operational_data/customers_stream")
```

Also we can infer the shema of the file ingested.


```
customers_df = spark.readStream
                    .format("cloudFiles")
                    .option("cloudFiles.format", "json")
                    .option("cloudFiles.useNotifications", "true")
                    .option("cloudFiles.schemaLocation",
                            "/Volumes/gizmobox/landing/operation_data/customers_stream/_schema")
                    .option("cloudFiles.inferColumnTypes", "true")
                    .load("/Volumes/gizmobox/landing/operational_data/customers_stream")
```

Schema evolution mode enabled

```
customers_df = spark.readStream
                    .format("cloudFiles")
                    .option("cloudFiles.format", "json")
                    .option("cloudFiles.useNotifications", "true")
                    .option("cloudFiles.schemaLocation",
                            "/Volumes/gizmobox/landing/operation_data/customers_stream/_schema")
                    .option("cloudFiles.inferColumnTypes", "true")
                    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
                    .load("/Volumes/gizmobox/landing/operational_data/customers_stream")
```
