## Streaming Ingestion from kafka and event hubs

In many real world scenarios, data doesn't arrive as files. Instead, it's generated continuously as *events*

User activity on website, transaction happening in a system, logs and telemetry data coming from IoT devices.
All of these are examples of event-based data.

*event-based data* are typically handled by systems like apache's kafka and azure event hubs(known as event streaming platforms)

In Databricks we use the kafka conector with Spark structured streaming directly or Spark declarative pipelines

- The key difference compared to autoloader is this.
With kafka or event hubs we ingest events continuosly as they are generated

```
df = spark.readStream
          .format("kafka")
          .option("kafka.bootstrap.servers", "host:port")
          .option("suscribe", "topic_name")
          .option("startingOffsets", "earliest")
          .load()
```

- **kafka.bootstrap.servers** especifies the Kafka server or the cluster that we are connecting to.
- **subscribe**: specifies the topic in the kafka server that we want to read the data from
- **startingOffsets**: controls where we start reading the data from in the continuous stream. With *earliest*, Spark reads all available data from the stream. With *latest* option, it reads only new data as it arrives.

In some cases we neet to convert or parse this data into a string before applying further transformations

```
df_parsed = df.selectExpr("CAST(value AS STRING)")
```

then we can write the tranfomr data to the data lake house typically in a delta table in a bronze layer.
