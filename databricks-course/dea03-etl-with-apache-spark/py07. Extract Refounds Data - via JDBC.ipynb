#### 1. Read Returns Data via JDBC

```python
dbutils.secrets.help()
```

```python
dbutils.secrets.listScopes()
```

```python
dbutils.secrets.list('dbricks-course')
```

```python
db_table = dbutils.secrets.get('dbricks-course', 'dbtable')
url_db = dbutils.secrets.get('dbricks-course', 'url-db')
password_db = dbutils.secrets.get('dbricks-course', 'passworddb')
username_db = dbutils.secrets.get('dbricks-course', 'user-db')
```

```python
df = (
    spark.read.format('jdbc')
        .option('url', url_db)
        .option('dbtable', db_table)
        .option('user', username_db)
        .option('password', password_db)
        .load()
)
display(df)
```

```python
df.writeTo('gizmobox.bronze.py_refunds').createOrReplace()
```
