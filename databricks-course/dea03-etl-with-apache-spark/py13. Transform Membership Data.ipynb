# Transform Memberships Data
1. Extract customer_id from the file path
2. Write transformed data to the Silver schema

```python
df = spark.table('gizmobox.bronze.v_memberships')
display(df)
```

## 1. Extract customer_id from the file path



```python
from pyspark.sql import functions as F

df_extrated_customer = (
    df.
    select(
        F.regexp_extract('path', '.*/([0-9]+)\\.png$', 1).cast('bigint').alias('customer_id'),
        F.col('content').alias('membership_card')
    )
)

display(df_extrated_customer)
```

## 2. Write transformed data to the Silver schame

```python
df_extrated_customer.writeTo('gizmobox.silver.py_memberships').createOrReplace()
```
